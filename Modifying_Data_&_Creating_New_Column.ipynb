```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('data/cleaned_ecommerce_data.csv')
df.head()
```

```python
# creating a new column using existing columns
# as, total_price = quantity * unit_price
df['total_price'] = df['quantity'] * df['unit_price']
df.head()
```

```python
# Vectorized Operations
df['unit_price'] = df['unit_price'] * 2
df.head()
```

### apply() Function
- apply() is used when you need to run a custom function row-wise or column-wise. Unlike vectorized df['col'] * 2, apply() is for logic/if-else
- apply() is used a lot for feature engineering
- But apply() is slower

```python
# Example 1: apply() on a column - Create price category
# First define the custom fuction which you want to apply()
def price_category(price: float) -> float:
    if price > 60000:
        return "Expensive" # always use return statement for apply()
    elif price > 20000:
        return "Mid"
    else: 
        return "Cheap"

# Now, Creating new column and implementing apply()
df['price_level'] = df['unit_price'].apply(price_category)
df.head()
```

```python
# Example 2: apply() with lambda on a column - Extract email domain
df['email_domain'] = df['email'].apply(lambda x: x.split('@')[1] if isinstance(x,str) and '@' in x else None)
df.head()
```

```python
# Example 3: apply() on Rows - axis=1 _Create discount based on multiple columns
def calc_discount(row):
    if row['status'] == 'Pending' and row['payment_method'] == 'Bank Transfer':
        return row['total_price'] * 0.10 # 10% discount
    else:
        return 0
df['discount'] = df.apply(calc_discount, axis=1) # axis=1 means row-wise, by default axis=0 that is column-wise which means one value at a time
df.head()
```

### .map() Function
- map() in pandas = "dictionary look-up for 1 column"
- It's used when we want to replace/transform each value in a single column using a distionary, function, or Series

```python
city_codes = {'Islamabad' : "ISL", 'Sialkot' : 'SKT', 'Multan' : 'MUL', 'Rawalpindi' : 'RWL', 'Karachi' : "KAR"}
df['city_code'] = df['city'].map(city_codes)
df.head()
```

### np.where()

```python
# we can also np.where()
df['bulk_order'] = np.where(df['quantity'] >= 3, 'Bulk', 'Small')
df.head()
```

### np.select()
- "Multiple if-elif-else in 1 line" for pandas columns
- map() is for 1-to-1 dictionary. np.select() is for when you have conditions
- Syntax:
np.select(condition list, choice list, default)

```python
conditions = [
    df['unit_price'] > 50000,
    df['unit_price'] > 20000,
    df['unit_price'] > 5000,
    df['unit_price'] <= 5000
]

choices = [
    'Premium',
    'High',
    'Mid',
    'Budget'
]

df['price_tier'] = np.select(conditions, choices, default='Unknown')
df.head()
```

### NOTE:

| Need           |  Use        |
|----------------|--------------
|Dict replacement|  map()      |
|2 conditions    |  np.where() |
|3+ conditions   | np.select() |
