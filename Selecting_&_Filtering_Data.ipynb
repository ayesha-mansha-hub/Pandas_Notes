```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/iris_data.csv",header=None, names=[
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
])
print(df.head())
```

```python
# selecting a specific column
print(df['sepal_length'])
print()
print(df['sepal_width'].head()) # we can also perform all functions on a specific column or columns
```

```python
# selecting multiple columns
print(df[['sepal_length','sepal_width']].head()) # pass a list of column names inside df[]
```

```python
# to select rows there are two functions
# -> .loc[] that is location which is the custom indices set by us  -> .iloc[] that is index location which is default position from 0,1,2,3,..
```

```python
# we can set a specific column as our index using .set_index()
df = df.set_index('sepal_length')
print(df.head())
```

```python
print(df.loc[4.6]) # selecting one row
print()
print(df.loc[[4.7,4.8]]) # selecting multiple rows
```

```python
# using .iloc[] -> iloc stands for index location
print(df.iloc[4])
print()
print(df.iloc[[4,7,8]])
```

```python
# index slicing
print(df.iloc[4:10])
```

```python
# slicing a section
print(df.iloc[4:12][["sepal_width","species"]])
```

```python
# NOTE:
# we can perform slicing using .loc[] but in iris dataset, the column I set as custom index contains NON Unique values spread-out in the entire dataset
# so .loc[] slicing behaves unreliably.
# Also, in .loc[] slicing 'end' is inclusive and in .iloc[] slicing, end is exclusive
```

## Boolean Filter

```python
# we can apply boolean filters on columns/Series
print(df["sepal_width"] > 3.5)
```

```python
df[df["sepal_width"] > 3.5] # it applies filter to entire Dataframe
```

```python
# combining comparison operators with logical operators in conditions of filter
# NOTE: In Python: it was "and", "or" and "not" 
# But in Pandas: "and" -> &, "or" -> |, "not" -> ~
print(df[(df["petal_width"] > 2.3) & (df["petal_length"] > 5.0)])   # AND
```

```python
print(df[(df['species'] == 'Iris-virginica') | (df["petal_length"] == 6.1)])
```

```python
print(df[~((df["species"] == 'Iris-virginica') & (df["petal_length"] < 5.0))])
```

```python
# Useful Filtering Method
print(df[df["species"].isin(["Iris-setosa","Iris-versicolor"])])
```

```python
print(df[df["petal_length"].between(4.8,5.0)]) # Inclusive range
```

```python
print(df[df["species"].str.contains("versicolor")])
```
