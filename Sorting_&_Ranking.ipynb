### Sorting

```python
# Sorting physically reorders your rows or columns, 
# while ranking assigns numerical positions to values while preserving the DataFrame's original index order
# After analyzing data, you usually want the extremes — the highest, lowest, best, or worst performing records,
# not the whole table in whatever order it happened to load in.
df.sort_values(by="quantity")                        # ascending, default
```

```python
df.sort_values(by="quantity", ascending=False).head()       # descending
```

```python
df.sort_values(
    by=["order_date", "quantity"],
    ascending=[True, False]
).head()
# With multiple columns, the first is the primary sort, the second only breaks ties within it 
# order_date groups stay primary, and only within each order_date does quantity sort high to low.
```

### NOTE:
- sort_values() returns a new DataFrame by default
- the original stays untouched unless you pass inplace=True or reassign it: df = df.sort_values(...).

## Ranking
- ranking = giving a position/order number to values in a column
- .rank() outputs a new Series/column
- index order is preserved
- tie handling through (min, average, max, dense, first) methods
##### NOTE: a tie occure when two or more records share the exact same value in the column you are sorting or ranking
### Handling Ties in Pandas `.rank()`
A **tie** occurs when two or more records share the exact same value in the column you are ranking. How pandas handles this tie depends on the `method` parameter passed to `.rank()`:
* **`average` (Default):** Averages the tied positions. If two entries tie for 2nd and 3rd place, both receive `(2 + 3) / 2 = 2.5`.
* **`min`:** Assigns the lowest rank to all tied items (`2`). The next person gets rank `4` (skips `3`).
* **`max`:** Assigns the highest rank to all tied items (`3`). The next person gets rank `4`.
* **`dense`:** Like `min`, but does not skip rank numbers. Both get rank `2`, and the next person gets rank `3`.
* **`first`:** Breaks ties by row sequence in the dataset. Whoever appears first gets `2`, and the second gets `3`.


```python
# Example 1: rank orders by unit_price - Most expensive first
df['price_rank'] = df['unit_price'].rank(ascending=False, method='min')
df
```

### Top & bottom records

```python
df.nlargest(7, "unit_price") 
```

```python
df.nsmallest(5, "price_rank")
```

nlargest()/nsmallest() are faster than sort_values().head() on big datasets — they never sort the rows you don't need.
#### NOTE:
We can also do Group-wise ranking _ groupby().rank()
Group-wise ranking evaluates values independently inside each category, resetting the rank counter (1, 2, 3...) whenever the group changes.
