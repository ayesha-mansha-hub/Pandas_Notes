### Creating Columns/Series

```python
import numpy as np
import pandas as pd
```

```python
# Pandas work with dataFrames (a labeled table _ rows & columns)
# it has different datatype per column
# a column is called "Series"
# a dataFrame has row indexes and column names
```

```python
a =[1,2,3,4,5,6,7] 

b = pd.Series([40000, 65000, 50000])                      # default 0,1,2 index
c = pd.Series([40000, 65000, 50000], index=["Ayesha","Ezza","Hurmat"]) # custom index
d = pd.Series({"Aman": 40000, "Eman": 65000})                    # dict keys become the index
s = pd.Series(a)
print(b,"\n")
print(c,"\n")
print(d,"\n")
print(s,"\n")
```

```python
array = np.arange(200,211,dtype=np.int32)
print(array)
print()
series = pd.Series(array,name='IDs') # name is the column/Series name
print(series)
```

### DataFrames

```python
# we can create DataFrame using pd.DataFrame()
# pd.DataFrame(data, index, columns, dtype, copy)
# we can pass data in different formats as:
# a list, a list of lists, a 2D NumPy array, a dictionary of lists, a list of dictionaries, a dictionary of dictionaries, a list of tuples
```

```python
l = [1,2,3,4,5]
df = pd.DataFrame(l)
print(df)
print()
df1 = pd.DataFrame(l, columns=['A']) # 1D list is considered as a column/series
print(df1)
```

```python
l = [[1,2,3,4],[5,6,7,8]]
df = pd.DataFrame(l) # here inner lists are considered as rows
print(df)
```

```python
array = np.array([[33,44,55],[66,77,88]])
df = pd.DataFrame(array, columns=['A','B','C'])
print(df)
```

```python
d = {"Ayesha" : 99, "Ezza" : 92, "Hurmat" : 88}
try: 
    df = pd.DataFrame(d)
    print(df)
except ValueError as e:
    print("DataFrame not created!", e) # because scalar dictionary values require an explicit index to create a Dataframe

print()
# But if we pass the above scalar dictionary as a list pandas will be able to recognize it
df1 = pd.DataFrame([d]) # in DataFrame "Keys" become "Column Names"
print(df1)
```

```python
d = {
    "Name" : ["Ayesha","Ezza", "Nadia"],
    "Marks" : [99, 98, 100]
}
df = pd.DataFrame(d)
print(df)
print()
data = {
    "Ayesha" : {"Math" : 99, "Physics" : 100},
    "Nadia" : {"Math" : 89, "Physics" : 96, "Class" : "BS"}
}
df = pd.DataFrame(data)
print(df)
list_of_dict = [{"Name" : "Ayesha", "RollNo" : 3001, "Class" : "BS"},{"Name" : "Momna" , "RollNo" : 3004}]
df1 = pd.DataFrame(list_of_dict)
print("\n",df1)
```

### Basic Inspection

```python
# Never start analyzing right after loading, Read the data first!
print(df.head()) # -> shows first five rows    
```

```python
print(df.tail()) # last five rows
```

```python
print(df.shape) # tells shape of data -> rows * cols
```

```python
print(df.info()) # useful to see how many Null/missing values
```

```python
print(df.describe()) # tells statistics/numerical distribution  _by default focuses on numerical columns
```

```python
print(df.sample()) #
```

```python
print(df.dtypes) # to see datatype of each column
```

```python
print(df.isnull()) # .isnull() tells where the missing values are in the dataframe
print()
print(df.isna()) # same as .isnull()
print(df.notnull()) # opposite of .isnull()
print(df.isnull().sum()) # it tells how many missing values each column has
```
